## **Grupo 1**



```python
import pandas as pd
df3= pd.read_csv("grupo1.csv")
df3.head()
```

```python
#exploración inicial de los datos
display(df3.describe())
```

```python
#Histograma de Calificaciones
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
df3['Calificacion'].hist(bins=10, edgecolor='black')
plt.title('Distribución de Calificaciones')
plt.xlabel('Calificación')
plt.ylabel('Frecuencia')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Modelo de Regresión Lineal
from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr
import numpy as np

# Preparar los datos
X = df3['HorasEstudio'].values.reshape(-1, 1)
y = df3['Calificacion'].values.reshape(-1, 1)

# Entrenar el modelo de regresión lineal
model = LinearRegression()
model.fit(X, y)

# Obtener la ecuación de la línea de regresión
intercept = model.intercept_[0]
slope = model.coef_[0][0]

print(f"Ecuación de la línea de regresión: Calificacion = {slope:.2f} * HorasEstudio + {intercept:.2f}")
```

```python
# Calcular el coeficiente de correlación de Pearson
correlacion_coeficiente, _ = pearsonr(df3['HorasEstudio'], df3['Calificacion'])

print(f"Coeficiente de correlación de Pearson: {correlacion_coeficiente:.2f}")
```

```python
#Gráfico de Dispersión con Línea de Regresión
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.regplot(x=df3['HorasEstudio'], y=df3['Calificacion'], scatter=True, color='blue', line_kws={'color': 'red'})

plt.title('Horas de Estudio vs. Calificación con Línea de Regresión')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificación')
plt.grid(True)
plt.show()
```

```python
# Mostrando el resultado del coeficiente de correlación de Pearson calculado anteriormente
print(f"El coeficiente de correlación de Pearson entre Horas de Estudio y Calificación es: {correlacion_coeficiente:.2f}")
```
