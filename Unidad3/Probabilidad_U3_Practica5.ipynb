## **Programa 5: Grafica de distribucion de probabilidad**

```python
#-----------------------------------------
#Grafica de una distrbucion de probabilidad discreta
#Con media, varianza y desviacion estandar
#--------------------------------------------------
import matplotlib.pyplot as plt
import numpy as np

#Datos: valores de la variable aleatoria
x= np.array([5,6,7,8,9,10])
p= np.array([0.05, 0.10, 0.20, 0.30,0.25, 0.10])
#Calculo del valor esperado(media)
esperanza=np.sum(x*p)
#Calculo de la esperanza al cuadrado
esperanza_cuadrado=np.sum(x**2*p)
#Calculo de la varianza y desviacion estandar
varianza = esperanza_cuadrado - esperanza**2
desviacion = np.sqrt(varianza)
```

```python
#Mostrar los resultados en consola
print(f"E(x)= {esperanza:.2f}")
print(f"E(X^2)= {esperanza_cuadrado:.2f}")
print(f"Varianza= {varianza:.2f}")
print(f"Desviacion estandar= {desviacion:.2f}")
```

```python
#Crear la grafica tipo 'stem' (tallos y hojas)
plt.stem(x, p, basefmt=" ", linefmt='b-', markerfmt='bo')
plt.title("Distribucion de probabilidad Discreta\ncon media y Desviacion estandar")
plt.xlabel("Valores de la variable aleatoria x")
plt.ylabel("p(x)")
plt.grid( True, linestyle="--", alpha=0.6)
#Dibujar lina vertical en la media
plt.axvline(x=esperanza, color="r", linestyle="--", label=f"Media(E(X))={esperanza:.2f}")

#Resaltar el rango de +-1 desviacion estandar
plt.axvspan(esperanza-desviacion, esperanza+desviacion, color='red', alpha=0.2, label=f"+-1 Desviacion estandar (o = {desviacion:.2f})")

#Mostrar la leyenda y limites
plt.legend()
plt.xlim(min(x) - 1, max(x) + 1) # Adjust x-axis limits to cover the data range
plt.xticks(x)
#Mostrar la grafica
plt.show()
```
