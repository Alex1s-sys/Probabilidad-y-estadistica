```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

plt.figure(figsize=(6,4))

# Parametros de la distribucion normal (ya definidos previamente, los redefinimos para este calculo)
media = 70 # Media
desv_std = 4 # Desviacion estandar
total_estudiantes = 500

# Calculamos la probabilidad de que un estudiante pese menos de 72 kg (P(X < 72))
prob_menos_72kg = norm.cdf(72, media, desv_std)

# Calculamos la probabilidad de que un estudiante pese menos de 68 kg (P(X < 68))
prob_menos_68kg = norm.cdf(68, media, desv_std)

# La probabilidad de que un estudiante pese entre 68 y 72 kg es P(X < 72) - P(X < 68)
prob_entre_68_y_72kg = prob_menos_72kg - prob_menos_68kg

# Calculamos el numero de estudiantes que pesan entre 68 y 72 kg
estudiantes_entre_68_y_72kg = total_estudiantes * prob_entre_68_y_72kg

print(f"Probabilidad de que un estudiante pese entre 68 y 72 kg: ",prob_entre_68_y_72kg)
print(f"Numero de estudiantes que pesan entre 68 y 72 kg: ",estudiantes_entre_68_y_72kg)
```
