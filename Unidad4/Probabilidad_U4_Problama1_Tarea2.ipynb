## **Problema 1: distribucion poisson y binomial**

### Análisis del Problema

#Tipo de Distribución:** Poisson

**Explicación:**
Es una distribución de Poisson porque estamos midiendo el número de ocurrencias de un evento (ingreso de personas) en un intervalo de tiempo específico (un día). Además, conocemos la media que es de 5 personas por día.



```python
import numpy as np
from scipy.stats import poisson
import matplotlib.pyplot as plt

# Parámetro lambda (media)
mu=5

# a) Probabilidad de que x sea igual a 2
prob_a=poisson.pmf(2, mu)
print(f"a) La probabilidad de que ingresen exactamente 2 personas es: {prob_a:.4f} ({prob_a*100:.2f}%)")

# b) Probabilidad de que x exceda de 10

prob_b=1-poisson.cdf(10, mu)
print(f"b) La probabilidad de que x exceda de 10 es: {prob_b:.4f} ({prob_b*100:.2f}%)")

# Explicación del inciso b

print("\nExplicación: Es poco probable que x exceda de 10, ya que la probabilidad es menor al 5%.")
```
