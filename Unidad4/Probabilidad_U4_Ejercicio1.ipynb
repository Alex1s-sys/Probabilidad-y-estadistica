## **Ejercicio 1: Distribucion normal**


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
#Generar datos para la distrubucion normal
media=70 #Media
desc_std=4 #Desviacion estandar
num_datos = 500

#Generamos los datos de manera aleatoria y los ponemos en una lista(array)
datos = np.random.normal(media, desc_std, num_datos)
datos[0:50] #Mostrar los primeros 50
```

```python
#Crear un rango de valores para crear una densidad
x= np.linspace(media-3*desc_std, media + 3* desc_std, num_datos)
#Calcular la PDF de la distribucion normal
pdf=1/ (desc_std * np.sqrt(2*np.pi)) * np.exp(-(x-media)**2/ (2 * desc_std **2))

#Para no imprimr toda la lista de valores(500) desplegamos los primero 50
print("Imprimir los primeros 50 Valores de x", x[0:50], " , y los primeros 50 pdf", pdf[0:50])
```

```python
#Graficar histograma
plt.figure(figsize=(6,4))
plt.hist(datos, bins=15, alpha=0.4, color='b') #Histograma de los datos
plt.title('Dsitrbiucion normal')
plt.xlabel('Peso en kg de los estudiantes')
plt.grid(True)
plt.show()
```
