### Análisis del Problema

#Tipo de Distribución:** Distribucion binomial

**Explicación:**
Hay un numero fijo de 1000 motores, ademas la posibilidad de que un motor salga defectuso no afecta a los demas motores y ademas la posibilidad de que un motor sea defectuoso es de 0.1 siendo significativamente baja

```python
from scipy.stats import binom

# Parámetros del problema
n=1000  # Número total de motores en el embarque
p=0.01  # Probabilidad de que un motor sea defectuoso

# Calcular la probabilidad de que ninguno sea defectuoso (k=0)
k_0=0
prob_0_defectuosos = binom.pmf(k_0, n, p)
print(f"Probabilidad de que ninguno sea defectuoso: {prob_0_defectuosos:.6f}")

# Calcular la probabilidad de que tres sean defectuosos (k=3)
k_3=3
prob_3_defectuosos=binom.pmf(k_3, n, p)
print(f"Probabilidad de que tres sean defectuosos: {prob_3_defectuosos:.6f}")

# Calcular la probabilidad de que cuatro sean defectuosos (k=4)
k_4=4
prob_4_defectuosos=binom.pmf(k_4, n, p)
print(f"Probabilidad de que cuatro sean defectuosos: {prob_4_defectuosos:.6f}")
```
