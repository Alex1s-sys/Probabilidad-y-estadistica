```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

#1. Cauntos estudiantes pesan mas de 79kg
plt.figure(figsize=(6,4))

#Parametros de la distribucion normal
media=70 #Media
desv_std=4 #Desviacion estandar
total_estudiantes=500

#VARIABLE para calculo de la probabilidad de que un estudiante que pese menos de 75 kg
peso_limite=75

#Primero nececitamos calcular la probabilidad de que un estudiante pese menos
#De 75kg utilizando la funcion de ditribucion acumulada (CDF) de la distribucion normal
prob_menos_75kg=norm.cdf(peso_limite,media,desv_std)
```

```python
#Calculo del numero de estudiantes que pesan menos de 75 kg
estudiantes_menos_75=total_estudiantes*prob_menos_75kg

print("Probabilidad de que un estudiante pese menis de 75kg: ", prob_menos_75kg)
print("Numero de estudiantes que pesan menos de 75kg: ", estudiantes_menos_75)

#Graficar la distribucion normal
x=np.linspace(media - 4 * desv_std, media + 4 * desv_std, 1000)
y=norm.pdf(x, media, desv_std)

plt.plot(x,y, 'b', linewidth=2)
plt.fill_between(x,y, where=(x>=peso_limite), color='red', alpha=0.5)
plt.title(r'Probabilidad de que un estudiante pese menos de 75kg. \n $\mu=70$, $\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```

```python
#2.Cuantos estudiantes pesan menos de 62kg

plt.figure(figsize=(6,4))

#Parametros de la distribucion normal
media=70 #Media
desv_std=4 #Desviacion estandar
total_estudiantes= 500

#Variable para calculo de la probabilidad de que un estudiante pese menos de 75 kg
peso_limite=62
#Primero nececitamos calcular la probabilidad de que un estudiante pese menos
#de 75kg utilizando la funcion de distribucion acumulativa (CDF)
prob_menos_62kg= norm.cdf(peso_limite, media, desv_std)

#Calculo del numero de estudiantes que pesan menos de 75kg
estudiantes_menos_62kg= total_estudiantes * prob_menos_62kg

print("Probabilidad de que un estudiante pese menos de 62 kg:", prob_menos_62kg)
print("Numero de estudiantes que pesan menos de 75 kg:", estudiantes_menos_62kg)
```

```python
#Calcular la distribucion normal
x=np.linspace(media - 4 * desv_std, media + 4 * desv_std,total_estudiantes)
y=norm.pdf(x, media, desv_std)

#Graficar la distribucion normal
plt.plot(x,y, 'b', linewidth=2)
plt.fill_between(x,y, where=(x>=peso_limite), color='skyblue', alpha=0.5)
plt.title('Probabilidad de que un estudiante pese menos de 62kg, \n $\mu=70$, $\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```

```python
#3. Cuantos estudiante pesan mas de 79kg
#2.Cuantos estudiantes pesan menos de 62kg

plt.figure(figsize=(6,4))

#Parametros de la distribucion normal
media=70 #Media
desv_std=4 #Desviacion estandar
peso_limite=79 #Peso limite
total_estudiantes= 500

#Variable para calculo de la probabilidad de que un estudiante pese menos de 75 kg
peso_limite=62
#Calculamos la probabilidad de que un estudiante pese menos o igual a 79 kg
prob_menos_79kg= norm.cdf(peso_limite, media, desv_std)
#La probabilidad de que un estudiante pese mas de 79kg es complementaria
prob_mas_79kg= 1 - prob_menos_79kg
#Calculo del numero de estudiantes que pesan mas de 79kg
estudiantes_mas_79kg= total_estudiantes * prob_mas_79kg

print("Probabilidad de que un estudiante pese menos de 79 kg:", prob_mas_79kg)
print("Numero de estudiantes que pesan menos de 79 kg:", estudiantes_mas_79kg)
```

```python
#Calcular la funcion de desnidad de probabilidad de la distribucion normal
x=np.linspace(media - 4 * desv_std, media + 4 * desv_std,total_estudiantes)
y=norm.pdf(x, media, desv_std)

#Graficar la distribucion normal
plt.plot(x,y, 'b', linewidth=2)
plt.fill_between(x,y, where=(x>=peso_limite), color='skyblue', alpha=0.5)
plt.title(r'Probabilidad de que un estudiante pese menos de 62kg, \n $\mu=70$, $\sigma=4$, X=500')
plt.xlabel('Peso (kg)')
plt.ylabel('Densidad de probabilidad')
plt.grid(True)
plt.show()
```
